```python
import pandas as pd
import numpy as np
```

```python
# Шаг 1 - выгрузка информации о продажах
df = pd.read_csv('/Users/mikhailstroganov/project_tms2/blinkit_orders.csv')
df
```

```python
df['promised_delivery_time'] = pd.to_datetime(df['promised_delivery_time'])
df['actual_delivery_time'] = pd.to_datetime(df['actual_delivery_time'])
df['order_date'] = pd.to_datetime(df['order_date'])
df['order_dt'] = df['order_date'].dt.date
df
```

```python
# Шаг 2 - (консультация с заказчиком) выполнение подсчетов согласно техническому заданию

# Метрики: кол-во заказов в день, сумма продаж в день, распределение выручки по payment_method

df_gr = df.groupby('order_dt', as_index=False).agg({'order_id':'count', 'order_total':'sum'})
df_piv = df.pivot_table(index=['order_dt'], columns=['payment_method'], values=['order_total'], aggfunc=['sum']).fillna(0)
```

```python
# шаг 3 - выгрузка Excel-файла

df_gr.to_excel('/Users/mikhailstroganov/project_tms2/blinkit1.xlsx', index=False)
df_piv.to_excel('/Users/mikhailstroganov/project_tms2/blinkit2.xlsx')
```

```python
writer = pd.ExcelWriter("blinkit.xlsx")

df_gr.to_excel(writer, sheet_name="Динамика продаж 2 ")
df_piv.to_excel(writer, sheet_name="Распределение продаж 2")

writer.close()
```
